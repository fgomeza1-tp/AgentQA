# Análisis de Datos

Este notebook contiene ejemplos de análisis exploratorio de datos y validación de datos utilizando Python.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo para gráficos
sns.set(style='whitegrid')

# Cargar datos de ejemplo
data = pd.read_csv('../dbt/seeds/example_seed.csv')

# Mostrar las primeras filas del conjunto de datos
data.head()
```

```python
# Análisis de datos

# Descripción estadística
data.describe()
```

```python
# Visualización de datos
plt.figure(figsize=(10, 6))
sns.histplot(data['column_name'], bins=30, kde=True)
plt.title('Distribución de column_name')
plt.xlabel('column_name')
plt.ylabel('Frecuencia')
plt.show()
```
